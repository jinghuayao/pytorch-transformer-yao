```python
# Simulating a batch of images: [Batch=10, Channels=3, Height=32, Width=32]
import torch
images = torch.randn(10, 3, 32, 32)

# Swap Channels (dim 1) and Height (dim 2)
swapped = images.transpose(1, 2)

print(swapped.shape)  # torch.Size([10, 32, 3, 32])
```
