```python
def causal_mask(size):
    mask = torch.triu(torch.ones((1, size, size)), diagonal=1).type(torch.int)
    return mask == 0
```

```python
causal_mask(3)
```

```python
def causal_mask_simple(size):
    mask = torch.tril(torch.ones((1, size, size)), diagonal=0).type(torch.int)
    return mask == 1
```

```python
causal_mask_simple(3)
```

```python
causal_mask_simple(3) == causal_mask(3)
```
